# 安裝PhotonVision

Reference:
    PhotonVision: [Quick Install](https://docs.photonvision.org/en/latest/docs/quick-start/quick-install.html)

Limelight/ Orange Pi/ Raspberry Pi: 可以透過[官方GitHub](https://github.com/PhotonVision/photonvision/releases/tag/v2025.3.1)給的.img.xz檔再用[BalenaEtcher](https://etcher.balena.io/#download-etcher)[燒到LimeLight](https://docs.limelightvision.io/docs/docs-limelight/getting-started/limelight-3#4-updating-limelightos)上之後就好了

Jeteson Orin/ 其他Linux Edge Computing Device: 也是到[官方GitHub](https://github.com/PhotonVision/photonvision/releases/tag/v2025.3.1)抓Linux Arm64/Linux x64的jar檔抓下來跑

一般的電腦(Windows PC/ MacOS): 還是到[官方GitHub](https://github.com/PhotonVision/photonvision/releases/tag/v2025.3.1)抓符合格式的jar檔

```{Remember} ⚠️ Remember
如果要裝在機器上，除了Limelight有針對突然斷電做優化，不然一般的邊緣裝置(例如Jetson Orin)還是在機器上另外拉獨立的電會比較好

電腦端可以先裝IP scanner到時候接到比較乾淨的網路(例如小黑)就比較好找到裝置

## 執行PhotonVision

- 如果是用BalenaEtcher燒.img.xz:通常通電開機之後就會自動跑了
- 如果是跑jar檔:下載之後跑 ``java -jar {file}.jar`` (記得裝Java JDK 17)
- 觀察端：
    - 用IP連接: ``{裝置IP}:5800``
    - 用mDNS連接(建議): ``{裝置名稱}:5800``

如果是接機器網路的話，裝置IP就會是10.TE.AM.IP(例如:``http://10.85.69.87:5800``), 不過要記得他要用固定IP不然之後要Debug會很麻煩

如果是用mDNS的話, 用 ``sudo nmtui`` 去設定hostname再重啟應該就可以用了(例如``http://jetson.local:5800``)

# GUI簡介

![PhotonVision截圖](https://photonvision.org/assets/demo-BbQQNrm0.png)

這是PhotonVision的GUI觀察介面，在右下角的Input框框中，如果有Auto Exposure(自動曝光)跟Auto White Balance(自動白平衡)能開就盡量開，這樣比較不容易因為比賽場地的燈光突然出狀況就爆掉；下面的Resolution跟Stream Resolution就是辨識解析度跟直播解析度，Resolution越高的話，他就能用更高的解析度找AprilTag(通常也能看更遠)，不過對於系統資源的消耗也會越大；Stream Resolution的話就盡量越低越好(降低網路負載也降低在CameraServer看的時候掉幀的機率(不過要給Driver看的話也不能太低就是了))

而中間的AprilTag的那個框框基本上不用動他，讓他保持原來的設定就好了

最右邊的那個3D是在相機校正過後才會有的，Targets會寫出每個偵測的AprilTag目標的狀態(Raw, Pitch, Yaw或(x,y,z, rotation, etc.))

Types(類別)的話可以自己去摸摸看
